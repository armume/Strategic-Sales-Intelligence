```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set(style="whitegrid")
```

```python
df = pd.read_csv(r'C:\Users\Ari\Desktop\DATA SET\retail-sales-analysis\1. data\train.csv')
df.head()
```

2. Convert dates, change format type, and remove null values

```python
df["Order Date"] = pd.to_datetime(df["Order Date"],dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"],dayfirst=True)
```

```python
df["Postal Code"] = df["Postal Code"].astype("string")
```

```python
df = df.dropna(subset=["Postal Code"])
```

3. Remove irrelevant columns.

```python
cols_drop = ["Row ID", "Customer Name", "Product Name", "Order ID", "Product ID"]
df = df.drop(columns=cols_drop)
```

4. Create new columns useful for analysis.

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Day"] = df["Order Date"].dt.day
```

```python
df["Shipping_Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
```

5. Handle outliers.

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Sales_Capped"] = df["Sales"].clip(lower, upper)
```
